```python
import pandas as pd
import numpy as np 

#2

a=pd.read_csv("/Users/zakariareeren/Desktop/numerique/notebooks/data/objects-on-the-moon.csv")
print(a.shape)
a.head(10)

#3

b=a.drop(columns="Unnamed: 0") 

#4 

b.info()
# La colonne Size est entièrement vide

#5

b.dropna(axis=1,how="all",inplace=True)
b

#Le inplace =True permet de modifier directement le tableau au lieu de creer une copie modifiée; 
#le axis=1 pour regarder les colonnes
# le how permet de choisir la methode, ici on supprime les colonnes dont toutes les cases sont vides. 

#6 

b.loc[88]

# cette ligne est vide

b.dropna(axis=0,how="all",inplace=True)
# axis=0 pour traiter des lignes 

b

# 7 

b.dtypes

# la colonne masse contient des chaines de caracteres au lieu de nombres. 

#8 

pd.Series(b["Mass (lb)"]).unique()

# Toutes les masses sont entières, et certaines cases contiennent des < > qui vont rendre la conversion en nombre difficile "


#9 

b=b.rename(columns={"Mass (lb)":"Mass (lb) orig"})

# Pour changer le nom de la colonne Mass afin de la conserver

b["Mass (lb)"]=pd.to_numeric(b["Mass (lb) orig"],errors="coerce")

# le errors=coerce permet d'écrire NaN lors des erreurs

b.dtypes # on a bien converti la colonne, qui contient maintenant des float

b["Mass (lb)"].isna().sum() #la console affiche np.int64(4) donc 4 NaN

# 10 

# D'abord reperer quelles sont les lignes où il y a un < ou un >
```
